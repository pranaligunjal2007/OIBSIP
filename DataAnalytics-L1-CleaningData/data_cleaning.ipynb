```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing values handled successfully!")
```

```python
print("Missing values after handling:")
print(df.isnull().sum())
```

```python
duplicates_before = df.duplicated().sum()

print("Duplicate rows before removal:", duplicates_before)
```

```python
df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicate rows after removal:", duplicates_after)
print("Rows after duplicate removal:", len(df))
```

```python
print("Current data types:")
print(df.dtypes)
```

```python
print("\nUnique values in categorical columns:")

print("\nSex:")
print(df["Sex"].unique())

print("\nEmbarked:")
print(df["Embarked"].unique())
```

```python
# Remove unnecessary spaces
df["Sex"] = df["Sex"].str.strip().str.lower()
df["Embarked"] = df["Embarked"].str.strip().str.upper()
df["Name"] = df["Name"].str.strip()

print("Text formatting standardized successfully!")

print("\nSex values:", df["Sex"].unique())
print("Embarked values:", df["Embarked"].unique())
```

```python
numeric_columns = [
    "PassengerId", "Survived", "Pclass",
    "Age", "SibSp", "Parch", "Fare"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print("Numeric datatypes corrected!")
print(df.dtypes)
```

```python
numeric_for_outliers = ["Age", "Fare", "SibSp", "Parch"]

outlier_summary = []

for column in numeric_for_outliers:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = ((df[column] < lower_bound) | 
                (df[column] > upper_bound)).sum()
    
    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outliers": outliers
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```
